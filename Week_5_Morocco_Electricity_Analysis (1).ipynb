```python
import pandas as pd

df = pd.read_csv('powerconsumption.csv')

df.rename(columns={df.columns[0]: 'Datetime'}, inplace=True)

df['Datetime'] = pd.to_datetime(df['Datetime'])

print(df.shape)
```

```python
Q1_3 = df['PowerConsumption_Zone3'].quantile(0.25)
Q3_3 = df['PowerConsumption_Zone3'].quantile(0.75)

IQR_3 = Q3_3 - Q1_3

lower_limit_3 = Q1_3 - 1.5 * IQR_3
upper_limit_3 = Q3_3 + 1.5 * IQR_3

outliers_3 = df[
    df['PowerConsumption_Zone3'] > upper_limit_3
]

print("Number of potential outliers:", len(outliers_3))

outliers_3[['Datetime', 'PowerConsumption_Zone3']].head()
```

```python
percentage = (len(outliers_3) / len(df)) * 100

print("Percentage of potential outliers:", percentage)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

environmental_corr = df[[
    'Temperature',
    'Humidity',
    'WindSpeed',
    'PowerConsumption_Zone1',
    'PowerConsumption_Zone2',
    'PowerConsumption_Zone3'
]].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    environmental_corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Between Environmental Factors and Electricity Consumption')
plt.tight_layout()
plt.show()
```
